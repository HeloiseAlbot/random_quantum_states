```python
import math 
import numpy as np
import matplotlib.pyplot as plt

n = 2
# Define the functions
def f1(t, n):
    return t * np.arccos(1 / np.sqrt(n))  # Example function

def f2(t, n):
    return np.pi / 2 - t * np.arccos(1 / np.sqrt(n))  # Example function

def f3(t, n):
    return t * np.arcsin(1 / np.sqrt(n-1))  # Example function

def f4(t):
    return np.pi / 4 * (2 - t)  # Example function

# Define the range for t
t_space = np.linspace(0, 1, 500)  # 500 points from 0 to 1

# Evaluate the functions
y1 = np.array([f1(t, n) for t in t_space])
y2 = np.array([f2(t, n) for t in t_space])
y3 = np.array([f3(t, n) for t in t_space])
y4 = np.array([f4(t) for t in t_space])

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(t_space, y1, label="$f_1(t) = t \\cdot \\arccos(1/\\sqrt{10})$", color="blue")
plt.plot(t_space, y2, label="$f_2(t) = \\pi/2 - t \\cdot \\arccos(1/\\sqrt{10})$", color="red")
plt.plot(t_space, y3, label="$f_3(t) = t \\cdot \\arcsin(1/\\sqrt{9})$", color="green")
plt.plot(t_space, y4, label="$f_4(t) = \\pi/4 (2 - t)$", color="purple")

# Add labels, legend, and grid
plt.xlabel("t", fontsize=14)
plt.ylabel("f(t)", fontsize=14)
plt.title("Comparison of Four Functions", fontsize=16)
plt.legend(fontsize=12)
plt.grid(True, linestyle="--", alpha=0.7)

# Show the plot
plt.show()
```
